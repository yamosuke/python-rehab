```python
a = [[0, 1],[2, 3]]
b = [[3, 2], [1, 0]]
c = []

for row in range(len(a)):
    tmp_list = []
    for col in range(len(a[0])):
        tmp_list.append(a[row][col] + b[row][col])
    c.append(tmp_list)
print(c)
```

```python
a = [[0, 1],[2, 3]]
b = [[3, 2], [1, 0]]
c = []

for a_items, b_items in zip(a, b):
    tmp_list = []
    for a_items, b_items in zip(a_items, b_items):
            tmp_list.append(a_items + b_items)
    c.append(tmp_list)
    
print(c)
```
