```python
# 例外処理
while True:
    try:
        price = int (input("金額を入力して下さい。"))
        if price < 0:
            raise ValueError ("金額は0（ゼロ）以上で入力して下さい。")
        count = int(input("人数を入力して下さい。"))
        if count < 1:
            raise ValueError ("人数は2人で入力して下さい。")
        print("金額//人数 = 一人当たり金額は", price // count, "円")
        break
    except ValueError:
        print("整数の値を入力して下さい。")
    except ZeroDivisionError:
        print("人数はゼロ以上を入力して下さい。")
    except Exception as e:
        print(e)
    finally:
        print("-" * 40)
```

```python
while True:
    try:
        num = int(input("正の整数を入力して下さい。"))
        if num <= 0:
            raise ValueError()
    except ValueError:
        print("数字ではないか、0以下の入力になっています。\nもう一度入力して下さい。")
    else:
        print(num)
        break
```

```python
while True:
    try:
        num = int(input("正の整数を入力して下さい。"))
        if num <=0:
            raise ValueError()
    except ValueError:
        print("数字ではないか、0以下の入力になっています。\nもう一度入力して下さい。")
    else:
        is_prime = True
        if num == 1:
            is_prime = False
        else:
            for i in range(2, num):
                if num % i == 0:
                    is_prime = False
                    break
        if is_prime:
            print(f"入力された「{num}」は素数です。")
        else:
            print(f"入力された「{num}」は素数ではありません。")
        break
```
