```python
# 小問題2　3つの数の最大値を返す関数（max() は使わずに）
def max_of_three(a, b, c):
        result = a
        for n in [b, c]:
            if n > result:
                result = n
        return result
    
print(max_of_three(44, 33, 22))
print(max_of_three(-15, -45, -2))
```

```python
#小問題4　["OK", "NG", "OK", "OK", "NG"] から不良率（%）を計算する
dims = ["OK", "NG", "OK", "OK", "NG","OK", "NG", "NG", "NG", "NG",] 

def defect(value):
    ng = 0
    for x in value:
        if x == "NG":
            ng +=1
    d_rate = (ng/len(value))*100
    return d_rate

rate = defect(dims)
print(f"不良率は {rate:.1f}% です。")
```

```python
# 小問題6 品番→在庫数の辞書を作って、出庫処理をする関数を書く（在庫が足りなければエラーメッセージ）
stock = {
    "A-001": 50, "A-002": 32, "A-003": 18, "A-004": 75, "A-005": 0,
    "A-006": 44, "A-007": 12, "A-008": 90, "A-009": 27, "A-010": 63,
    "A-011": 5,  "A-012": 38, "A-013": 81, "A-014": 20, "A-015": 56,
    "A-016": 9,  "A-017": 47, "A-018": 33, "A-019": 100, "A-020": 15,
    "A-021": 68, "A-022": 24, "A-023": 3,  "A-024": 52, "A-025": 41,
    "A-026": 19, "A-027": 86, "A-028": 30, "A-029": 7,  "A-030": 59,
}

code = input("品番を指定して下さい。")
num = int(input("何個出荷するか数量を入力して下さい。"))

def stocks(code, num):
    if code in stock :
        if stock[code] < num:
            print(f"在庫数量が足りません。(在庫：{stock[code]}、出荷{num}個)")
        else:
            stock[code] -= num
            print(f"品番コード：{code} の在庫残数は{stock[code]}です。")
    else:
        print("品番が見つかりません")
        return

stocks(code, num)
```

```python
# 小問題8 単価と数量のリストから、合計の購買金額を計算する
orders = [
    ("A-001", 120, 10),
    ("A-002", 80, 25),
    ("A-003", 300, 4),
    ("A-004", 45, 100),
    ("A-005", 1500, 2),
]

total = 0

for code, price, qty in orders:
    amount = price * qty
    print(f"品番：{code}, 単価：{price:>5,}円, 数量：{qty:>3}個, 金額：{amount:>6,}円")
    total += amount
    
print(f"合計金額は {total:,} 円です。")
```

```python
# 文章の中の単語の出現回数を、辞書で数える

text = "the quick brown fox jumps over the lazy dog the fox".split(" ")
words = set(text)
for word in words:
    count = 0
    for word2 in text:
        if word == word2:
            count +=1
        else:
            continue
    print(f"［{word}］の出現回数は{count}回です。")
```
