```python
# これは行列に対する積
x = np.array(
    [[1, 2, 3],
    [4, 5, 6]]
)

result = x * 3
print(x)
print(result)
```

```python
# これは行列同士の足し算
x = np.array(
    [[1, 2, 3],
    [4, 5, 6]]
)

y = np.array(
    [[10, 20, 30],
     [40, 50, 60]
     ]
)

result = x + y
print(result)
```

```python
print(x.shape)
print(y.shape)
```

```python
# 行列同士の積は 1個目の行列の列数と2個目の行列の行数が一致しないといけない。
x = np.array(
    [[1, 2, 3],
    [4, 5, 6]]
)

y = np.array(
    [[10, 20, 30],
     [40, 50, 60]
     ]
)

print(x.shape)
print(y.shape)

result = np.dot(x, y)
print(result)
```

```python
x = np.array(
    [[1, 2, 3],
    [4, 5, 6]]
)

y = np.array(
    [[10, 20],
     [30, 40],
     [40, 50]]
)

print(x.shape)
print(y.shape)

result = np.dot(x, y)
print(result)
```
