ブールインデックス

```python
a = np.array([10, 20, 30, 40, 50])
```

```python
print(a[a > 25])
print(a[a % 20 == 0])
```

```python
arr = np.array([5, 12, 18, 25, 33, 40])
```

```python
print(arr[(arr >= 20) & (arr <= 40) ])
```

```python
print(arr[(arr<10) | (arr>30)])
```

```python
print(arr[~(arr == 25) & ~(arr == 40)])
```

```python
a = np.arange(12).reshape(3,4)
print(a)
```

```python
print(a[a > 5])
```

```python
print(a[:, 1] > 2)  # これではbooleanの値しか表示しない
```

```python
a2 = a[a[:, 1] >= 6]  # この行内には条件に合致する値無し
print(a2)
```
