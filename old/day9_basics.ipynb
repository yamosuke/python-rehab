```python
import csv
catalog = [("hat", 2000), ("shirt", 1000), ("socks", 500), ("pants", 500)]
with open("catalog.csv", "w", encoding="UTF-8", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["item", "price"])  # 見出し行
    writer.writerows(catalog)
```

```python
with open("catalog.csv", encoding="UTF-8") as file:
    for row in csv.DictReader(file):
        print(row["item"], row["price"])
```

```python
with open("catalog.csv", encoding="UTF-8") as file:
    for row in csv.DictReader(file, fieldnames=["item", "price"]):
        print(row["item"], row["price"])
```

```python
import csv
product_size = [("A-001", 10.02), ("A-002", 9.98) , ("A-003", 10.11), ("A-004",9.95), ("A-005",10.00)]
with open("product_size.csv", "w", encoding="UTF-8", newline="") as size_file:
    writer = csv.writer(size_file)
    writer.writerow(["code", "size(mm)"])
    writer.writerows(product_size)
    
with open("product_size.csv", encoding="UTF-8") as file:
    for row in csv.DictReader(file):
        print(row["code"], row["size(mm)"])
```

```python
with open("product_size.csv", encoding="UTF-8") as file:
    count = 0
    total = 0
    for row in csv.DictReader(file):
        total += float(row["size(mm)"])
        count += 1
    
    avg = total/count
    
    print(avg, "(mm)")
```
