```python
# フォルダ内のファイル一覧を作って、CSVに書き出す
# 対象：指定したフォルダ
# CSVの列　ファイル名、拡張子、サイズ、更新日時
# 出力先： file_list.csv
```

```python
from pathlib import Path
from datetime import datetime
import csv

folder = Path(input("フォルダのパスを入力して下さい。").strip('"'))
try:
    with open("file_list.csv", "w", encoding="utf-8-sig", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["ファイル名", "拡張子", "ファイルサイズ", "更新日"])  
        
        for p in folder.iterdir():
            try:
                if p.is_file():
                    file_name = p.name
                    suffix_name = p.suffix
                    pst = p.stat()
                    file_size = pst.st_size
                    dt = datetime.fromtimestamp(pst.st_mtime)
                    formated_dt = dt.strftime("%Y/%m/%d %H:%M")
                    writer.writerow([file_name, suffix_name, file_size, formated_dt])
            except PermissionError:
                print(f"読めないファイルをスキップ： {p.name}")
                continue
            
except FileNotFoundError:
    print("入力したパスにファイルが存在しません。")
except NotADirectoryError:
    print("入力したパスからファイル名を除いて下さい。")
except PermissionError:
    print("file_list.csv を閉じてから実行して下さい")
else:
    print("file_list.csv に書き出しました。")
```
