```python
import numpy as np
```

```python
x = np.array(
    [[10.02, 12.05, 14.66],
     [20.23, 22.52, 24.78]]
)

y= np.array(
    [[1, 2, 3],
     [4, 5, 6]
    ]
)

z = np.array([[1, 2],
              [3, 4],
              [5, 6]])
```

```python
# 足し算
result1 = x + 1
print(result1)

print()

result2 = x + y
print(result2)
```

```python
# 引き算
result1 = x - 1
print(result1)

print()

result2 = x - y
print(result2)
```

```python
# 掛け算
result1 = x * 2
print(result1)

print()

result2 = x * y
print(result2)

print()

result3 = x @ z
print(result3)
```

```python
# 合計
x = np.array(
    [[10.02, 12.05, 14.66],
     [20.23, 22.52, 24.78]]
)

r = np.sum(x)
print(r)

r = np.sum(x, axis=0)
print(r)

r = np.sum(x, axis=1)
print(r)
```

```python
# 標準
x = np.array(
    [[10.02, 12.05, 14.66],
     [20.23, 22.52, 24.78]]
)

r = np.mean(x)
print(x)

print()

r2 = np.mean(x, axis=0)
print(r2)
```

```python
x = np.arange(12).reshape(3, 4)
print(x)
```

```python
print(np.sum(x))
print()
print(np.sum(x, axis=0))
```

```python
print(np.mean(x))
print(np.mean(x, axis=1))
```

```python
x = np.array(
    [[10.02, 12.05, 14.66],
     [20.23, 22.52, 24.78]]
)
```

```python
print(np.max(x))
print(np.min(x))
```

```python
print(np.std(x))
```

```python
print(np.round(x))
```

```python
# 寸法の標準、標準偏差、中央値

size = np.random.rand(4,3)
print(size)
print()

print(np.mean(size))
print()
print(np.std(size))
print()
print(np.median(size))
```
