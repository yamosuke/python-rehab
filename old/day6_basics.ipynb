```python
def odd_even(n):
    if n%2 == 0:
        print(f"入力された数字「{n}」は偶数です。")
    else:
        print(f"入力された数字「{n}」は奇数です。")
```

```python
print(odd_even(2))
print(odd_even(5))
```

```python
odd_even(2)
odd_even(5)
```

```python
def odd_even(n):
    return "奇数" if n%2 else "偶数"

n = input("input number: ")
if n.isdigit():  # isdigit() で数字かどうかを判定
    n = int(n)  # その後でintに変換
    print(f"入力した数字「{n}」は「{odd_even(n)}」です。")
else:
    print("error: 整数の数字を入力して下さい。")
```

```python
def average(*values):
    return sum(values)/len(values)
```

```python
average(10.02, 9.98, 10.11)
sizes = [10.02, 9.98, 10.11, 9.95, 10.00]
average(*sizes)
```

```python
def show(*args, **kwargs):
    print(args, type(args))
    print(kwargs, type(kwargs))
    
show(10, 20, 30, unit="mm", lot="L-01")
```

```python
def count(*t, **d):
    for i, x in enumerate(t, 1):
        print("[", i, "]", x)
    for i, (k, v) in enumerate(d.items(), len(t)+1):
        print("[", i, "]", k, ":", v)
```

```python
count("hotcake", "pizza", snack="parfait", dinner="steak")
```
