```python
# 規格値 10.00 ± 0.05 とする

rng = np.random.default_rng(0)  # 乱数生成
data = rng.normal(10.0, 0.02, size=(100)) # 平均10.0、標準偏差0.02の正規分布でデータを作成
lower, upper = 9.95, 10.05

ave = np.mean(data)  # サイズの平均値
print(ave)

sd = np.std(data, ddof=1) # サイズの標準偏差
print(sd)
```

```python
# 工程能力指数 CP
cp = (upper - lower) / (6 * sd)
print(cp)
```

```python
# 工程能力指数　Cpk

cpk = min((upper - ave) / (3 * sd), (ave - lower) / (3 * sd))
print(cpk)
```

```python
# 工程能力判定
def process_capability(cpk):
    if cpk >= 1.33:
        print("工程能力は非常に優れています。")
    elif cpk >= 1.00 and cpk < 1.33:
        print("工程能力は十分です。")
    elif cpk < 1.00:
        print("工程能力が不足しています。")

print(f"工程能力指数： {cpk :.5f}")
process_capability(cpk)
```
